# Module 01: Chip Security Landscape — Lab

This interactive lab explores the security landscape through classification exercises and analysis of real-world attack patterns.

**Objectives:**
1. Classify known attacks into the taxonomy (passive/active, invasive/non-invasive)
2. Analyze attack complexity and impact
3. Map attacks to relevant NIST standards and countermeasures

```python
import json
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Define real-world attacks with attributes
attacks = [
    {"name": "Spectre", "year": 2018, "type": "Passive",
     "category": "Speculative Execution", "invasiveness": "Non-invasive",
     "impact": "High", "affected": ["Intel", "AMD", "ARM"],
     "cve": "CVE-2017-5753"},
    {"name": "Meltdown", "year": 2018, "type": "Passive",
     "category": "Speculative Execution", "invasiveness": "Non-invasive",
     "impact": "Critical", "affected": ["Intel"],
     "cve": "CVE-2017-5754"},
    {"name": "Rowhammer", "year": 2014, "type": "Active",
     "category": "Fault Injection", "invasiveness": "Non-invasive",
     "impact": "High", "affected": ["DDR3", "DDR4"],
     "cve": "CVE-2015-0565"},
    {"name": "Heartbleed", "year": 2014, "type": "Passive",
     "category": "Software Vulnerability", "invasiveness": "Non-invasive",
     "impact": "Critical", "affected": ["OpenSSL"],
     "cve": "CVE-2014-0160"},
    {"name": "Power Analysis (CPA)", "year": 2005, "type": "Passive",
     "category": "Side-Channel", "invasiveness": "Non-invasive",
     "impact": "High", "affected": ["Smart Cards", "Embedded"],
     "cve": "N/A"},
    {"name": "Voltage Glitching", "year": 2017, "type": "Active",
     "category": "Fault Injection", "invasiveness": "Semi-invasive",
     "impact": "High", "affected": ["MCU", "SoC"],
     "cve": "N/A"},
    {"name": "FIB Probing", "year": 1990, "type": "Active",
     "category": "Physical Probe", "invasiveness": "Invasive",
     "impact": "Critical", "affected": ["Smart Cards", "ASIC"],
     "cve": "N/A"},
    {"name": "Timing Attack", "year": 1996, "type": "Passive",
     "category": "Side-Channel", "invasiveness": "Non-invasive",
     "impact": "Medium", "affected": ["RSA", "AES"],
     "cve": "N/A"},
    {"name": "Cache侧信道", "year": 2016, "type": "Passive",
     "category": "Side-Channel", "invasiveness": "Non-invasive",
     "impact": "High", "affected": ["Intel", "AMD", "ARM"],
     "cve": "CVE-2016-8650"},
    {"name": "BadUSB", "year": 2014, "type": "Active",
     "category": "Supply Chain", "invasiveness": "Non-invasive",
     "impact": "High", "affected": ["USB Devices"],
     "cve": "N/A"},
]

df = pd.DataFrame(attacks)
print("Attack Database:")
print(df[['name', 'year', 'type', 'category', 'invasiveness', 'impact']].to_string(index=False))
```

```python
# Visualization: Attack taxonomy breakdown
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# 1. Attacks by type (Passive vs Active)
type_counts = df['type'].value_counts()
colors_type = ['#2ecc71', '#e74c3c']
axes[0].bar(type_counts.index, type_counts.values, color=colors_type)
axes[0].set_title('Attacks by Type', fontsize=14)
axes[0].set_ylabel('Count')

# 2. Attacks by invasiveness
inv_counts = df['invasiveness'].value_counts()
colors_inv = ['#3498db', '#f39c12', '#e74c3c']
axes[1].barh(inv_counts.index, inv_counts.values, color=colors_inv[:len(inv_counts)])
axes[1].set_title('Attacks by Invasiveness', fontsize=14)
axes[1].set_xlabel('Count')

# 3. Attacks over time
axes[2].scatter(df['year'], range(len(df)), c=[{'Passive': '#2ecc71', 'Active': '#e74c3c'}[t] for t in df['type']],
               s=100, zorder=5)
for i, row in df.iterrows():
    axes[2].annotate(row['name'], (row['year'], i), fontsize=8, ha='left', va='center',
                     xytext=(5, 0), textcoords='offset points')
axes[2].set_title('Timeline of Notable Attacks', fontsize=14)
axes[2].set_xlabel('Year')
axes[2].set_yticks([])

plt.tight_layout()
plt.show()

print(f"\nTotal attacks catalogued: {len(df)}")
print(f"Passive vs Active: {type_counts.get('Passive', 0)} / {type_counts.get('Active', 0)}")
```

```python
# Mapping attacks to NIST standards and countermeasures
standards_map = {
    "Side-Channel": {
        "NIST Standards": ["FIPS 140-3 (Level 3+)", "SP 800-175B"],
        "Countermeasures": ["Masking", "Hiding", "Threshold Implementation"],
        "Evaluation Method": "TVLA (SP 800-90B)"
    },
    "Fault Injection": {
        "NIST Standards": ["FIPS 140-3 (Level 4)", "Common Criteria EAL4+"],
        "Countermeasures": ["Redundancy Checks", "Error Detection Codes", "Voltage Monitoring"],
        "Evaluation Method": "Fault injection testing per ISO/IEC 17825"
    },
    "Speculative Execution": {
        "NIST Standards": ["SP 800-175B", "IR 8425"],
        "Countermeasures": ["Retpoline", "IBRS", "KPTI", "Microcode Updates"],
        "Evaluation Method": "Microarchitectural analysis"
    },
    "Supply Chain": {
        "NIST Standards": ["SP 800-161 Rev. 1", "FIPS 140-3"],
        "Countermeasures": ["Hardware Root of Trust", "Secure Boot", "Tamper Evidence"],
        "Evaluation Method": "Supply chain audit + CC evaluation"
    }
}

print("=" * 70)
print("ATTACK → STANDARD → COUNTERMEASURE MAPPING")
print("=" * 70)
for category, mapping in standards_map.items():
    print(f"\n{category}:")
    print(f"  NIST Standards:     {', '.join(mapping['NIST Standards'])}")
    print(f"  Countermeasures:    {', '.join(mapping['Countermeasures'])}")
    print(f"  Evaluation Method:  {mapping['Evaluation Method']}")

# Print which attacks in our database map to which categories
print("\n" + "=" * 70)
print("ATTACK CLASSIFICATION SUMMARY")
print("=" * 70)
for _, attack in df.iterrows():
    cat = attack['category']
    if cat in standards_map:
        countermeasures = standards_map[cat]['Countermeasures']
        print(f"{attack['name']:25s} → {cat:25s} → CM: {countermeasures[0]}")
    else:
        print(f"{attack['name']:25s} → {cat:25s} → CM: (varies)")
```
